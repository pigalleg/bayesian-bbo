# 3. Extending the baseline
Change `function_id` to 2–9 after placing the corresponding real initial data. Compare EI and UCB only with actual evaluations, and document the measured outcomes in later notebooks.

```python
from pathlib import Path
from bayesian_bbo import (
    CONFIG,
    BayesianOptimizer,
    OptimisationConfig,
    append_observation,
    format_submission,
    load_observations,
    remaining_budget,
)

function_id = 1
root = Path('../data')
observations = load_observations(root, function_id)
optimizer = BayesianOptimizer(
    observations,
    OptimisationConfig(
        budget=remaining_budget(observations, function_id),
        acquisition='ei',
        seed=CONFIG["random_state"],
    ),
)
if optimizer.remaining:
    point = optimizer.suggest()
    print(f'Function {function_id}:', format_submission(point))
else:
    print('Query budget exhausted; no new candidate.')
print('Best measured:', observations.best_input, observations.best_output)
# After evaluation: append_observation(root, function_id, point, actual_output, week=<current week>), then optimizer.observe(point, actual_output).
```
