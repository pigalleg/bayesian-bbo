```python
from pathlib import Path
from bayesian_bbo import BayesianOptimizer, OptimisationConfig, append_observation, format_submission, load_observations

root = Path('../data')
function_id = 1
observations = load_observations(root, function_id)
optimizer = BayesianOptimizer(observations, OptimisationConfig(budget=20, seed=42))
point = optimizer.suggest()
print('Submit this candidate:', format_submission(point))
print('Best measured output so far:', optimizer.history.best_output)
```
